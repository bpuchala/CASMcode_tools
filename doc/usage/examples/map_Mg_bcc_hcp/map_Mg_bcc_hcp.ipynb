# Map BCC Mg to HCP Mg

## Specify the "parent" and "child" structure

**This step:**

- Specify the parent structure (higher symmetry) and child structure (lower symmetry)
- For the parent we use a [xtal.Structure]
- For the child we use a [xtal.Structure]

```python
from libcasm.xtal.structures import BCC, HCP

parent = BCC(a=3.55, atom_type="Mg")
child = HCP(a=3.17, c=5.14, atom_type="Mg")
```

```python
import libcasm.xtal as xtal
import libcasm.configuration as casmconfig

xtal_prim = xtal.Prim.from_atom_coordinates(structure=parent)
parent_prim = casmconfig.Prim(xtal_prim)
```

```python
import pathlib
from casm.tools.map import (
    StructureMappingSearch,
    StructureMappingSearchOptions,
)

# Mapping search options
opt = StructureMappingSearchOptions(
    max_n_atoms=8,
)

# Merge new results with existing results if re-running
merge = True

# Results directory
results_dir = pathlib.Path("results")

# Run search
StructureMappingSearch(opt=opt)(
    parent=parent,
    parent_prim=parent_prim,
    child=child,
    results_dir=results_dir,
    merge=merge,
)
```

## Find best mapping - up to 8 atoms

```python
import libcasm.xtal as xtal
from casm.tools.map import StructureMappingSearchOptions, map_to_structure


# Mapping search options
opt = StructureMappingSearchOptions(
    max_n_atoms=8,
)

# Run mapping
mappings, uuids = map_to_structure(
    child=child,
    parent_structure=parent,
    opt=opt,
)
best_mapping = mappings[0]
```

```python
print(xtal.pretty_json(best_mapping.to_dict()))
```

## Find top 10 best mappings - up to 8 atoms

```python
import libcasm.xtal as xtal
import libcasm.configuration as casmconfig
from casm.tools.map import StructureMappingSearchOptions, map_to_structure
from casm.tools.map._StructureMappingSearch import (
    read_results,
    tabulate_results,
    write_results,
)


# Mapping search options
opt = StructureMappingSearchOptions(
    max_n_atoms=8,
    total_k_best=10,
    total_max_cost=0.3,
)

# Run mapping
mappings, uuids = map_to_structure(
    child=child,
    parent_structure=parent,
    opt=opt,
)
```

```python
# Display results table
tabulate_results(
    opt=opt,
    search_results=mappings,
    uuids=uuids,
    child=child,
    parent_prim=parent_prim,
)
```

```python
# Write results
options_history = []

write_results(
    search_results=mappings,
    uuids=uuids,
    parent=parent,
    child=child,
    parent_prim=parent_prim,
    results_dir=results_dir,
    options=opt,
    options_history=options_history,
)
```

```python
# Read results
parent_prim = None
parent_prim, mappings, uuids, options_history = read_results(
    results_dir=results_dir,
)
```
